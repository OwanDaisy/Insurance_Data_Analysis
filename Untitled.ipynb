### Exploratory Data Analysis

Firstly let's begin by exploring some test statistics for our quantitative variables to get some insight about our data

```python
df.describe()
```

```python
x = ["age","bmi","children","charges"]
for i in x:
    print(f"mean_{i}: {df[i].mean()}")
    print(f"median_{i}:{ df[i].median()}")
```

On initial view, these four attributes seem to be approximately symmetric and not heavily skewed as the mean and median are close in value, with no extreme outliers pulling the mean away from the center. This is a good point with regards to modelling.

let's visualize our distribution for more insight

```python
plt.hist(df.age, bins = 20)
plt.hist(df.bmi, bins = 20)
```

```python
plt.hist(df.charges, bins = 20)
```

The charges column is highly skeweed to the right, letting us know that exrtremely costly insurance charges are less common than extremely small ones. This makes it unlikely that the errors in the model will truly be centered at zero. Let us try transforming the scale

```python
#applying a log transform on the response variable
np.random.seed(0)
df["charges_log"] = np.log(df.charges)
plt.hist(df.charges_log)
```

```python
#applying a log transform on the response variable
np.random.seed(0)
df["charges_log2"] = np.log2(df.charges)
plt.hist(df.charges_log2)
```

```python
np.random.seed(0)
df["charges_sqrt"] = np.sqrt(df.charges)
plt.hist(df.charges_sqrt)
```

```python
np.random.seed(0)
np.random.seed(0)
df["charges_cbrt"] = np.cbrt(df.charges)
plt.hist(df.charges_cbrt)
```

The "most normal" distribution obtained seems to be from the log transformation......so we'll work with that

### Bivariate analysis / Feature Selection

Let's assess which variables impact our dependent variable(charges for the modelling)

```python
df_numeric = df[['age','bmi',"children","charges","charges_log","charges_log2"]]
df_numeric.corr()
```

```python
sb.heatmap(df_numeric.corr(),cmap='Reds', annot = True)
plt.show()
```

Initially, we observe that using a log2 or log 10 transformation doesn't make any difference. Additionally, the is a higher correlation between age and charges than any other numenic variable

```python
sb.pairplot(df_numeric, kind = "scatter", plot_kws = {'alpha' : 0.4})
plt.show()
```

interestingly, we can observe three distinct groups/clusters when plotting age and charges.......but no distinct pattern in bmi. These patterns could be due to the fact that the dataset si synthetic, or maybe bad sampling
